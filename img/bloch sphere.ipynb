```python
from qutip import Bloch

b = Bloch()
b.add_vectors([0, 0, 1])

b.render()

# transparent background
b.fig.patch.set_alpha(0)
b.axes.set_facecolor((1, 1, 1, 0))

# save the actual Bloch figure
b.fig.savefig(
    "bloch_sphere.png",
    dpi=600,
    transparent=True,
    bbox_inches="tight",
    pad_inches=0.05
)
```
